```python
# Added quotes around 'a' and 'b' so Python knows they are text
lists = [('a', 1), ('b', 1), ('a', 3), ('b', 4)]

key = {}

# Changed 'list' to 'item' so we don't break the built-in list() command
for item in lists:

    # Removed the broken .keys
    if item[0] not in key or item[1] > key[item[0]]:
        
        key[item[0]] = item[1]

# Since we aren't using a function, we just print the answer directly!
print(list(key.items()))
```
