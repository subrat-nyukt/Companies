```python
df_driver_transform["Age_group_new"]=None
df_driver_transform["Rev_bins_new"]=None
df_driver_transform["Truck_new"]=None
df_driver_transform["Truck_type_new"]=None
```

```python
df_driver_transform
```

```python
df_driver_transform["Age_group"].unique()
```

```python
for i in range(len(df_driver["Nationality"])):
  if df_driver_transform["Age_group"][i]=='25-34':
     df_driver_transform["Age_group_new"][i]=3
  elif df_driver_transform["Age_group"][i]=='35-44':
     df_driver_transform["Age_group_new"][i]=2
  
  else:
     df_driver_transform["Age_group_new"][i]=1
```

```python
df_driver_transform["Rev_bins"].unique()
```

```python
for i in range(len(df_driver["Nationality"])):
  if df_driver_transform["Rev_bins"][i]=='800-1200':
     df_driver_transform["Rev_bins_new"][i]=5
  elif df_driver_transform["Rev_bins"][i]=="1200-1600":
     df_driver_transform["Rev_bins_new"][i]=4
  elif df_driver_transform["Rev_bins"][i]=="400-800":
     df_driver_transform["Rev_bins_new"][i]=3
  elif df_driver_transform["Rev_bins"][i]=="0-400":
     df_driver_transform["Rev_bins_new"][i]=2

  else:
     df_driver_transform["Rev_bins_new"][i]=0
```

```python
df_driver_transform["Truck"].unique()
```

```python
for i in range(len(df_driver["Nationality"])):
  if df_driver_transform["Truck"][i]=='12M':
     df_driver_transform["Truck_new"][i]=2
  
  
  else:
     df_driver_transform["Truck_new"][i]=1
```

```python
df_driver_transform["Truck_type"].unique()
```

```python
for i in range(len(df_driver["Nationality"])):
  if df_driver_transform["Truck_type"][i]=='Flatbed':
     df_driver_transform["Truck_type_new"][i]=2
  
  
  else:
     df_driver_transform["Truck_type_new"][i]=1
```

```python
df_driver_transform
```

```python
df_driver_transform_v1=df_driver_transform.drop(columns=["Nationality","Age_group","Rev_bins","Truck","Truck_type"])
```

```python
df_driver_transform_v1
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
cols = ["Total_Revenue","No of orders","Avg Rev/Order"]
df_driver_transform_v1[cols]=scaler.fit_transform(df_driver_transform_v1[cols])
```

```python
df_driver_transform_v1
```

```python
from sklearn.cluster import KMeans

wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k, init="k-means++")
    kmeans.fit(df_driver_transform_v1)
    wcss.append(kmeans.inertia_)
plt.figure(figsize=(12,6))    
plt.grid()
plt.plot(range(1,11),wcss, linewidth=2, color="red", marker ="8")
plt.xlabel("K Value")
plt.xticks(np.arange(1,11,1))
plt.ylabel("WCSS")
plt.show()
```

```python
kmeans = KMeans(n_clusters = 5, init='k-means++')
clusters = kmeans.fit_predict(df_driver_transform_v1)
df_driver_transform_v1["label"] = clusters
```
