# Understanding df.shift() and Vectorized Comparisons in Pandas

This notebook breaks down exactly how to use `.shift()` to track changes over time (like log files, status changes, or time-series data).

```python
import pandas as pd

# 1. Create a practical sample dataset 
# Imagine this is a daily log of users and their account statuses
data = {
    'Date': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'User': ['Alice', 'Alice', 'Alice', 'Bob', 'Bob'],
    'Status': ['Active', 'Active', 'Suspended', 'Active', 'Active']
}
df = pd.DataFrame(data)

print("=== 1. ORIGINAL DATAFRAME ===")
display(df)
```

### What does `shift(1)` actually do mathematically?
It takes the entire DataFrame and pushes it down by 1 row. The top row gets filled with `NaN` (Not a Number/Null) because there is nothing to pull down from above.

```python
shifted_df = df.shift(1)

print("=== 2. WHAT df.shift(1) LOOKS LIKE ===")
display(shifted_df)
```

### Vectorized comparison for the entire DataFrame
`df != df.shift(1)` checks every single cell and asks: *"Is this value DIFFERENT from the value that was here yesterday (the row above)?"*

If it is different, it returns `True`.

```python
differences = df != df.shift(1)

print("=== 3. BOOLEAN DIFFERENCES (True means the value changed) ===")
display(differences)
```

### Practical Application: Finding Exact Moments of Change
In the real world, you use this to find the exact moment something broke or changed state.

We want to find where `Status` changed, BUT we only care if it's the same `User`. (We don't want it to flag as a "change" just because the row switched from Alice to Bob).

```python
status_changed = (df['Status'] != df.shift(1)['Status']) & (df['User'] == df.shift(1)['User'])

print("=== 4. REAL WORLD USE: Exactly when did a user's status change? ===")
# We use the boolean mask to filter the original dataframe
display(df[status_changed])
```
